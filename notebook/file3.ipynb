```python
import pandas as pd

file = r"data\T3_hlth_ehis_cd1e__custom_24003937_linear_2_0.csv"

df = pd.read_csv(file)
df.head(10)
```

```python
df.columns = df.columns.str.strip()

principali = [
    "Geopolitical entity (reporting)",
    "Health problems",
    "Sex",
    "Age class",
    "TIME_PERIOD",
    "OBS_VALUE"
]

completo = df[principali]

completo = df[principali]
inserire = [c for c in principali if c in df.columns]

completo = df[inserire]

dizionario = {
    "Geopolitical entity (reporting)": "Paese",
    "Health problems": "Problemi di saluite",
    "Sex": "Sesso",
    "Age class": "Fascia di eta",
    "TIME_PERIOD": "Anno di riferimento",
    "OBS_VALUE": "Valore osservato"
}

completo = completo.rename(columns=dizionario)

display(completo.head(10))
```

```python
# Selezione dei dati del 2019
dati = df[
    (df["TIME_PERIOD"] == 2019) &
    (df["Sex"] == "Females") &
    (df["Age class"] == "65 years or over")
]

# Creazione della tabella
tabella = dati.pivot_table(
    index="Health problems",
    columns="Geopolitical entity (reporting)",
    values="OBS_VALUE"
)

display(tabella.round(1))
```
