```python
import pandas as pd

file_path = r"data\T3.2_hlth_silc_05__custom_23921391_linear_2_0.csv"

df = pd.read_csv(file_path)
df.head(10)
```

```python
df.columns = df.columns.str.strip()

colonne_utili = ["Age class", "Sex", "Geopolitical entity (reporting)", "TIME_PERIOD", "OBS_VALUE"]
colonne_presenti = [c for c in colonne_utili if c in df.columns]

df_filtrato = df[colonne_presenti]

dizionario_nomi = {
    'Age class': 'Anno di nascita',
    'Sex' : 'Genere',
    'Geopolitical entity (reporting)': 'Paese', 
    'TIME_PERIOD': 'Anno di riferimento',
    'OBS_VALUE': 'Valore osservato'
}

df_filtrato = df_filtrato.rename(columns=dizionario_nomi)

if 'Valore osservato' in df_filtrato.columns:
    df_filtrato = df_filtrato.dropna(subset=['Valore osservato'])

display(df_filtrato.head(10))
```
