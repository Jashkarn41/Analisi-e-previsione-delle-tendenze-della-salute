```python
import pandas as pd

file_path = r"data\T5.1_hlth_hlye__custom_23928152_linear_2_0.csv"

df = pd.read_csv(file_path)
display(df)
```

```python
df.columns = df.columns.str.strip()

colonne_utili = ['Sex', 'hlth_hle', 'Geopolitical entity (reporting)', 'TIME_PERIOD', 'OBS_VALUE', 'ArithmeticError']
colonne_presenti = [c for c in colonne_utili if c in df.columns]

df_filtrato = df[colonne_presenti]

dizionario_nomi = {
    'Sex': 'Genere',
    'hlth_hle': 'Salute e benessere',
    'Health and life expectancy': 'Salute e aspettativa di vita',
    'Geopolitical entity (reporting)': 'Paese',
    'TIME_PERIOD': 'Anno di riferimento',
    'OBS_VALUE': 'Valore osservato',
    'ArithmeticError': 'Errore aritmetico'
}

df_filtrato = df_filtrato.rename(columns=dizionario_nomi)

if 'ArithmeticError' in df_filtrato.columns:
    df_filtrato = df_filtrato.drop(columns=['ArithmeticError'])

display(df_filtrato)
```
