```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
file_hlth = r'data\T5.1_hlth_hlye__custom_23928152_linear_2_0.csv'
file_proj = r'data\T5.2_proj_stp25__custom_23927919_linear_2_0.csv'
file_le = r'data\T5.3_tps00208__custom_23961916_linear_2_0.csv'

df_hlth = pd.read_csv(file_hlth)
df_proj = pd.read_csv(file_proj)
df_le = pd.read_csv(file_le)

for df in [df_hlth, df_proj, df_le]:
  df.columns = df.columns.str.strip()

colonne_utili_hlth = [
    'Geopolitical entity (reporting)',
    'Sex',
    'Health and life expectancy',
    'TIME_PERIOD',
    'OBS_VALUE',
]
df_hlth_filtrato = df_hlth[colonne_utili_hlth].copy()

diz_hlth = {
    'Geopolitical entity (reporting)': 'Paese',
    'Sex': 'Genere',
    'Health and life expectancy': 'Indicatore_salute',
    'TIME_PERIOD': 'Anno',
    'OBS_VALUE': 'Valore',
}
df_hlth_filtrato = df_hlth_filtrato.rename(columns=diz_hlth).dropna(
    subset=['Valore']
)

df_hlth_it = df_hlth_filtrato[df_hlth_filtrato['Paese'] == 'Italy']
display(df_hlth_it.head(10))

colonne_utili_proj = [
    'Geopolitical entity (reporting)',
    'Demographic indicator',
    'TIME_PERIOD',
    'OBS_VALUE',
]
df_proj_filtrato = df_proj[colonne_utili_proj].copy()

diz_proj = {
    'Geopolitical entity (reporting)': 'Paese',
    'Demographic indicator': 'Indicatore_demografico',
    'TIME_PERIOD': 'Anno',
    'OBS_VALUE': 'Valore',
}
df_proj_filtrato = df_proj_filtrato.rename(columns=diz_proj).dropna(
    subset=['Valore']
)

df_proj_le_it = df_proj_filtrato[
    (df_proj_filtrato['Paese'] == 'Italy')
    & (df_proj_filtrato['Indicatore_demografico'].str.contains('Life expectancy'))
]
display(df_proj_le_it)
```
