```python
import pandas as pd

file_path = r"data\T5.3_tps00208__custom_23961916_linear_2_0.csv"

df = pd.read_csv(file_path)
df.head(10)
```

```python
df.columns = df.columns.str.strip()

colonne_utili = ['Geopolitical entity (reporting)', 'Demographic indicator', 'TIME_PERIOD', 'OBS_VALUE', 'Sex', 'Age', 'Unit', 'Flag Codes', 'Health indicator']
colonne_presenti = [c for c in colonne_utili if c in df.columns]

df_filtrato = df[colonne_presenti].copy()

dizionario_nomi = {
    'geo': 'Codice_paese',
    'Geopolitical entity (reporting)': 'Paese',
    'Demographic indicator': 'Indicatore_demografico',
    'TIME_PERIOD': 'Anno',
    'OBS_VALUE': 'Valore',
    'Sex': 'Sesso',
    'Age': 'Età',
    'Unit': 'Unità',
    'Flag Codes': 'Codici_flag',
    'Health indicator': 'Indicatore_salute'
}

df_filtrato = df_filtrato.rename(columns=dizionario_nomi)

if 'Valore' in df_filtrato.columns:
    df_filtrato = df_filtrato.dropna(subset=['Valore'])

display(df_filtrato.head(10))
```
