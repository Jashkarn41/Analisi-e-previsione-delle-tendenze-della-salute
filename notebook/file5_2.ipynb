```python
import pandas as pd

file_path = r"data\T5.2_proj_stp25__custom_23927919_linear_2_0.csv"

df = pd.read_csv(file_path)
df.head(10)
```

```python
df.columns = df.columns.str.strip()

colonne_utili = ['Geopolitical entity (reporting)', 'TIME_PERIOD', 'OBS_VALUE']
colonne_presenti = [c for c in colonne_utili if c in df.columns]

df_filtrato = df[colonne_presenti]

dizionario_nomi = {
    'Geopolitical entity (reporting)': 'Paese',
    'TIME_PERIOD': 'Anno di riferimento',
    'OBS_VALUE': 'Valore osservato'
}

df_filtrato = df_filtrato.rename(columns=dizionario_nomi)

if 'Paese' in df_filtrato.columns:
    df_filtrato['Paese'] = df_filtrato['Paese'].str.strip()

display(df_filtrato.head(10))
```
